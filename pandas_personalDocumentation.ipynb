### Working with Series

We use the `pd.Series()` object constructor to convert an already existing array/list to a Series. It is not a function, it is a constructor

There are a few arguments that we can pass into the Series() constructor.
 - `index=[]` - This will give us indexing in the left-most column where serialisation appears by default (0, 1, 2, ...). We can now give these indices a label to read through. This `[]` can be anything - A numpy array, a list, a tuple, a dictionary.

<mark>The index parameter should contain the exact number of entires as the series. If not we are stomped in the face by a `ValueError`!</mark>

```python
data = [100.02, "A string inside numbers", 105] # to check what would happen if there are multiple data types 
series = pd.Series(data, index=["A", "B", "C"])
print(series)
```

### To Access a value directly within a series 
---

We use `series.loc` property.

`loc` stands for <mark>location by label</mark>

The `loc[]` property will return the elements present with that label. It is just like how lists are indexed and accessed, but a bit different in the syntax. Inside the `[]` we pass the label that we have defined. This idea is similar to how a dictionary stores its values.

If the paramter passed into `[]` does not exist, we get a `KeyError`

```python
print(series)
print(series.loc["A"])
```

### Changing the values in a Series
---

We use the `loc` property to access the element and then we assign it to a new value

```python
new_series = pd.Series(["Ryzen 9", "Ryzen 7", "Ryzen 5"], index=["CPU1", "CPU2", "CPU3"])
new_series.loc["CPU1"] = "AMD Ryzen 9 9950X3D"
print(new_series)
```

### `iloc` - Locating a value by integer position
---

The exact same way how indexed lists in python have their values extracted. The syntax is as follows:

`<series>.iloc[<integer_index>]`

```python
print(new_series)
print("*"*15)
print(new_series.iloc[1])
```

### Filtering by Labels in a Series
---

We use similar indexing/conditions and syntax as in NumPy to extract data that satisfy a condition.

```python
# using numpy to make a custom data set and then converting it to series
import numpy as np

DataSeries = pd.Series(np.array(np.arange(0, 250, 5)))

# Let us take that the condition is to print every integer here that has a value above or equal to 200
print(DataSeries[DataSeries >= 200])
```

## Pandas DataFrames
---


- A DataFrame is a 2-D data structure, similar to an Excel sheet

To create a DataFrame we use the `DataFrame()` constructor. Inside we pass our data structure and if needed, provide index/label values to it.

```python
data = {"Name": ['Spongebob', 'Patrick', 'Squidward'], "Age": [30, 35, 50]}

df = pd.DataFrame(data, index=["Character 1", "Character 2", "Character 3"]) # index to change the labels
print(df)
```

### Accessing Rows in DataFrames
---

To access individual rows in a DataFrame, we use the same `loc` property. We may also use `iloc` to access the data.

```python
print(df)
print("*"*15)

print(f"Using loc")
print(df.loc["Character 1"])
print("*"*15)
print(f"Using iloc")
print(df.iloc[2])
```

### To add a new Column
---

We use `df["<Column Title>"] = [<rows>]`.

The number of entries in `<rows>` should match with the number of entries already existing.

```python
print(df)
print("*"*15)

df["Job"] = ["Cook", "Jobless lol", "Cashier"]
print(df)
```

### To add a new Row
---

An easy way to add a new row is to create a new df, then concatenate it

`pd.concat([<df1>, <df2>, ...])`

Earlier, `append` was used, but nowadays it is not appreciated to use that. 

```python
print(df)
print("*"*15)

new_row = pd.DataFrame([{"Name" : "Sandy", "Age": 28, "Job": "Engineer"}], index=["Character 4"]) # Add as many dictionaries as the number of rows you want to add
df = pd.concat([df, new_row])
print(df)
```

```python
another_new_row = pd.DataFrame([{"Name" : "Mr. Crabs", "Age": 60, "Job": "Manager"}], index=["Character 5"])
df = pd.concat([df, another_new_row])
print(df)
```

### Setting a column as the label for easy access
---

It is much easier to work with labels than with indices. Therefore, we can set the name or any uniquely identifiable property from the column as the label so that we don't have to depend on the indices every time we want to aceess that row. Here is how we do that:
`df = pd.read_csv("<filename.csv>", index_col="<columun_name>")`

### Reading only specific columns for a given entry and slicing
---

Imagine that in the below code, we have to only extract the details for the Pokemon `charizard` and we only need the weight and the abilities, so in order to select specific columns for a given entry we do the following. We still use the `loc` property, but with a slight modification.

Syntax: `df.loc["<start_element>":"<end_element>", ["<property 1>", "<property 2>"]]`

```python
print(df.loc["Charizard", ["weight_kg", "abilities"]])

# Slicing 

print(df.loc["Charizard":"Blastoise", ["height_m", "weight_kg"]])
```

### Aggregate Functions in Pandas
---

Why use aggregate fuctions?
 - Reduces a set of values into a single summary value
 - We can use it to summarize and analyze data
 - Often used with the groupby() function {Note to self: this is similar to SQL...}

Common aggregate functions:
 - `df.mean()`
 - `df.sum()`
 - `df.min()`
 - `df.max()`
 - `df.count()`

Adding an extra keyowrd argument `numeric_only=True` will automatically make pandas take in only the numeric values in the data set and will ignore others. This helps in not getting any errors while operating on the entire dataset with these functions.

```python
# To find the mean of any field that is numeric in the dataset
print(df.mean(numeric_only=True))

# To find the sum of all the fields that are numeric
print(df.sum(numeric_only=True))

# To find the minumum and the maximum of all fields numeric
print(df.min(numeric_only=True))
print(df.max(numeric_only=True))

# To count the number of values in each column
print(df.count()) # count ignores empty fields
```

### Aggregate fucntions for a single column
---

To apply aggregate functions on specific columns, we use the following technique


```python
# To print the average height of all the pokemons
print(f"{df["height_m"].mean()} meters")

# To print the average weight of all the pokemons
print(f"{df["weight_kg"].mean()} kg")

# To find the sum of all the weights of certain pokemon
print(df.loc["Charmander":"Arcanine", ["weight_kg"]].sum())

```

## Data Cleaning
---

The process of fixing/removing incomplete, incorrect or unnecessary data


### Dropping columns using `df.drop()`
---

To remove columns, we use the `drop()` method. Syntax is as follows: `df.drop(columns=["<column1_name>", "<column2_name>"])`

```python
# Let's say we want to drop the criteria if a pokemon is a legendary pokemon or not :(
df = pd.read_csv("pokemon.csv", index_col="name")

# To remove the legenday pokemon from the dataframe, we do
df = df.drop(columns=["is_legendary"])
print(df)
```

### To handle missing data
---

We use the `df.dropna(subset=["<column1>", "<column2>"])` to drop all those rows that are missing a value

Here, `dropna` stands for `Drop notAvailable` (or) drop the items that are not available

```python
# let's say we want to drop the row if that row does not have a value for type2

df = df.dropna(subset=["type2"])
print(df.to_string())
```

### To Fill any field that is missing a value
---

We saw how we can remove those rows from the df where there are missing elements for a column, now instead of dropping that column, we fill it with a value that we want it to hold. So we use the `df.fillna({"<column_name>" : <value_we_want_it_to_hold>, ...})` to fill those columns.

Here, `fillna` stands for `Fill not Available` (or) fill all the columns that are empty

```python
df = pd.read_csv("pokemon.csv", index_col="name")
df = df.fillna({"type2" : "NONE"})

print(df.to_string())
```

### Fixing inconsistencies / inconsistent values
---

Let's say in here, we want to change all those types where it is spelled as `water` to `WATUH` or `grass` to `Grass`, we use the replace method:

`df["<column_name>"] = df["<column_name>"].replace({<Python dictionary with key value pairs>})`

```python
df["type1"] = df["type1"].replace({"grass" : "GRASS", "water": "WATUH"}) # Take a troll on the British accent lol
print(df.to_string())
```

### Standardise Text
---

A self-explanatory example given below

```python
df["japanese_name"] = df["japanese_name"].str.upper()
print(df.to_string())
```

### To Fix/Change Data Types
---

Here we wil be attempting to change the data type of the non-legendary to be `False` and legendary to be `True`. This is a conversion from `int16` to `bool`

To do this, we use the `.astype()` method. 

```python
df = pd.read_csv("pokemon.csv", index_col="name")
df["is_legendary"] = df["is_legendary"].astype(bool)
print(df.to_string())
```

### To remove the duplicate entries
---

The simplest line of code possibly...

Use the `drop_duplicates()` method on the data frame to automatically remove the duplicates

```python
df = pd.read_csv("pokemon.csv")
df = df.drop_duplicates()
print(df.to_string())
```
