Pandas has two ways to handle structured data. 
 - A DataFrame (A `DataFrame` is a 2-D Data Structure that can store different types of data in columns. Similar to how a spreadsheet works in Excel or how a DB is in SQL)
 - A Series (A linear 1-D Data Structure that can also store different types of data in columns. Similar to how a list looks like in Vanilla Python)

### Working with Series

We use the `pd.Series()` object constructor to convert an already existing array/list to a Series. It is not a function, it is a constructor

There are a few arguments that we can pass into the Series() constructor.
 - `index=[]` - This will give us indexing in the left-most column where serialisation appears by default (0, 1, 2, ...). We can now give these indices a label to read through. This `[]` can be anything - A numpy array, a list, a tuple, a dictionary.

<mark>The index parameter should contain the exact number of entires as the series. If not we are stomped in the face by a `ValueError`!</mark>

```python
data = [100.02, "A string inside numbers", 105] # to check what would happen if there are multiple data types 
series = pd.Series(data, index=["A", "B", "C"])
print(series)
```

### To Access a value directly within a series 
---

We use `series.loc` property.

`loc` stands for <mark>location by label</mark>

The `loc[]` property will return the elements present with that label. It is just like how lists are indexed and accessed, but a bit different in the syntax. Inside the `[]` we pass the label that we have defined. This idea is similar to how a dictionary stores its values.

If the paramter passed into `[]` does not exist, we get a `KeyError`

```python
print(series)
print(series.loc["A"])
```

### Changing the values in a Series
---

We use the `loc` property to access the element and then we assign it to a new value

```python
new_series = pd.Series(["Ryzen 9", "Ryzen 7", "Ryzen 5"], index=["CPU1", "CPU2", "CPU3"])
new_series.loc["CPU1"] = "AMD Ryzen 9 9950X3D"
print(new_series)
```

### `iloc` - Locating a value by integer position
---

The exact same way how indexed lists in python have their values extracted. The syntax is as follows:

`<series>.iloc[<integer_index>]`

```python
print(new_series)
print("*"*15)
print(new_series.iloc[1])
```

### Filtering by Labels in a Series
---

We use similar indexing/conditions and syntax as in NumPy to extract data that satisfy a condition.

```python
# using numpy to make a custom data set and then converting it to series
import numpy as np

DataSeries = pd.Series(np.array(np.arange(0, 250, 5)))

# Let us take that the condition is to print every integer here that has a value above or equal to 200
print(DataSeries[DataSeries >= 200])
```

## Pandas DataFrames
---


- A DataFrame is a 2-D data structure, similar to an Excel sheet

To create a DataFrame we use the `DataFrame()` constructor. Inside we pass our data structure and if needed, provide index/label values to it.

```python
data = {"Name": ['Spongebob', 'Patrick', 'Squidward'], "Age": [30, 35, 50]}

df = pd.DataFrame(data, index=["Character 1", "Character 2", "Character 3"]) # index to change the labels
print(df)
```

### Accessing Rows in DataFrames
---

To access individual rows in a DataFrame, we use the same `loc` property. We may also use `iloc` to access the data.

```python
print(df)
print("*"*15)

print(f"Using loc")
print(df.loc["Character 1"])
print("*"*15)
print(f"Using iloc")
print(df.iloc[2])
```

### To add a new Column
---

We use `df["<Column Title>"] = [<rows>]`.

The number of entries in `<rows>` should match with the number of entries already existing.

```python
print(df)
print("*"*15)

df["Job"] = ["Cook", "Jobless lol", "Cashier"]
print(df)
```

### To add a new Row
---

An easy way to add a new row is to create a new df, then concatenate it

`pd.concat([<df1>, <df2>, ...])`

Earlier, `append` was used, but nowadays it is not appreciated to use that. 

```python
print(df)
print("*"*15)

new_row = pd.DataFrame([{"Name" : "Sandy", "Age": 28, "Job": "Engineer"}], index=["Character 4"]) # Add as many dictionaries as the number of rows you want to add
df = pd.concat([df, new_row])
print(df)
```

```python
another_new_row = pd.DataFrame([{"Name" : "Mr. Crabs", "Age": 60, "Job": "Manager"}], index=["Character 5"])
df = pd.concat([df, another_new_row])
print(df)
```
